```python
import nltk
nltk.download('wordnet')
from nltk.corpus import wordnet as wn


def similaridade_entre_palavras(p1, p2):
  sentidos1 = wn.synsets(p1)
  sentidos2 = wn.synsets(p2)
  if not sentidos1 or not sentidos2:
    return 0
  return sentidos1[0].wup_similarity(sentidos2[0]) or 0


texto1 = ['cat', 'sleeping', 'sofa']
texto2 = ['feline', 'resting', 'room']

# Calcular a similaridade média

similaridades = [similaridade_entre_palavras(p1, p2) for p1, p2 in zip(texto1, texto2)]
media = sum(similaridades) / len(similaridades)
print(f"Similaridade semântica aproximada: {media:.2f}")
```

```python
import nltk

nltk.download('punkt_tab')
nltk.download('stopwords')
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
from nltk.corpus import wordnet as wn

stop_words = set(stopwords.words('english'))

def similaridade_entre_palavras(p1, p2):
  sentidos1 = wn.synsets(p1)
  sentidos2 = wn.synsets(p2)
  if not sentidos1 or not sentidos2:
    return 0
  return sentidos1[0].wup_similarity(sentidos2[0]) or 0

print('Digite uma frase em inglês:')
texto1 = input()
print('Digite outra frase em inglês:')
texto2 = input()

# Tokenize the input phrases
tokens1 = word_tokenize(texto1)
tokens1 = [word for word in tokens1 if word not in stop_words]
tokens2 = word_tokenize(texto2)
tokens2 = [word for word in tokens2 if word not in stop_words]

# Calculate the average similarity between all pairs of words from the two phrases
similaridades = []
for token1 in tokens1:
  for token2 in tokens2:
    similaridades.append(similaridade_entre_palavras(token1, token2))

print(similaridades)
media = sum(similaridades) / len(similaridades) if similaridades else 0 # Avoid division by zero
print(f"Similaridade semântica aproximada: {media:.2f}")
```
