```python
c# Renan Antonio Hammerschmidt Krefta
# Victor Silva Camargo
# Vinícius Silva Camargo

lexico = ["abacate", "abacaxi", "abobora", "abobrinha", "ananás", "maça", "mamão",
"manga", "melancia", "melão", "mexerica", "morango"]

palavra = input("Palavra: ")

def lista_n_gram_unique(palavra, n):
  lista = []
  for i in range(len(palavra) - n + 1):
    if palavra[i:i+n] not in lista:
      lista.append(palavra[i:i+n])
  return lista

def interseccao(lista1, lista2):
  lista = []
  for i in range(len(lista1)):
    for j in range(len(lista2)):
      if lista1[i] == lista2[j]:
        lista.append(lista1[i])
  return lista

A = lista_n_gram_unique(palavra, 2)
print(A)

indice_similaridade = []

for i in range(0, len(lexico)):
  B = lista_n_gram_unique(lexico[i], 2)
  C = len(interseccao(A, B))
  indice_similaridade.append(2 * C / (len(A) + len(B)))

indice_similaridade, lexico = zip(*sorted(zip(indice_similaridade, lexico), reverse=True))
print(indice_similaridade[:5])
print(lexico[:5])
```
