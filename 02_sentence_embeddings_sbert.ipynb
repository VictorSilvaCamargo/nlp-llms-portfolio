```python
from sentence_transformers import SentenceTransformer, util
```

```python
model = SentenceTransformer('all-MiniLM-L6-v2')
```

```python
# Duas listas of sentenças
sentencas1 = ['The cat sits outside',
             'A man is playing guitar',
             'The new movie is awesome']

sentencas2 = ['The dog plays in the garden',
              'A woman watches TV',
              'The new movie is so great']

#Computa embedding
embeddings1 = model.encode(sentencas1, convert_to_tensor=True)
embeddings2 = model.encode(sentencas2, convert_to_tensor=True)

#Computa similaridade de cosine
cosine_scores = util.cos_sim(embeddings1, embeddings2)

#Pares de sentenças e seu score
for i in range(len(sentencas1)):
    print("{} \t\t {} \t\t Score: {:.4f}".format(sentencas1[i], sentencas2[i], cosine_scores[i][i]))
```

```python
# Exemplo comparativo com TF-IDF

# Listas de sentenças
sentencas1 = ['Shipment of gold damaged in a fire',
             'Delivery of silver arrived in a silver truck',
             'Shipment of gold arrived in a truck']

# Query
sentencas2 = ['gold silver truck',
              'gold silver truck',
              'gold silver truck']

# Calcula embedding das listas
embeddings1 = model.encode(sentencas1, convert_to_tensor=True)
embeddings2 = model.encode(sentencas2, convert_to_tensor=True)

# Calcula similaridade de cosseno
cosine_scores = util.cos_sim(embeddings1, embeddings2)

# Imprime score
for i in range(len(sentencas1)):
    print("{} \t\t {} \t\t Score: {:.4f}".format(sentencas1[i], sentencas2[i], cosine_scores[i][i]))
```

```python
model = SentenceTransformer('sentence-transformers/distiluse-base-multilingual-cased-v1')
```

```python
sentencas1 = ['O gato comeu o sapato',
             'O menino joga futebol com os amigos',
             'A aula do professor Emerson é sensacional']

sentencas2 = ['O cachorro brinca na grama',
              'A menina lê um livro no seu quarto',
              'Como foi incrível a aula do professor Emerson']

embeddings1 = model.encode(sentencas1, convert_to_tensor=True)
embeddings2 = model.encode(sentencas2, convert_to_tensor=True)

cosine_scores = util.cos_sim(embeddings1, embeddings2)

for i in range(len(sentencas1)):
    print("{} \t\t {} \t\t Score: {:.4f}".format(sentencas1[i], sentencas2[i], cosine_scores[i][i]))
```
