```python
from transformers import BertTokenizer, BertForNextSentencePrediction
import torch
```

```python
# Carregar o tokenizer e o modelo BERT pré-treinado em português
tokenizer = BertTokenizer.from_pretrained('neuralmind/bert-base-portuguese-cased')
model = BertForNextSentencePrediction.from_pretrained('neuralmind/bert-base-portuguese-cased')
```

```python
# Duas frases de exemplo em português
frase_1 = "Eu adoro programar em Python."
frase_2 = "É uma das linguagens de programação mais populares."

#frase_2 = "A comida estava deliciosa ontem"

# Tokenizar as frases (adicionar os tokens especiais necessários)
inputs = tokenizer(frase_1, frase_2, return_tensors='pt')

# Passar as frases tokenizadas pelo modelo
outputs = model(**inputs)

# A saída contém os logits, que são pontuações usadas para prever se a segunda frase é a continuação da primeira
logits = outputs.logits

# As pontuações: O índice [0] refere-se a "is_next" e o índice [1] refere-se a "not_next"
is_next_score = logits[0][0].item()
not_next_score = logits[0][1].item()

# Exibir os resultados
if is_next_score > not_next_score:
    print(f"A frase 2 pode ser continuação da frase 1 (Score: {is_next_score})")
else:
    print(f"A frase 2 talvez não seja a continuação da frase 1 (Score: {not_next_score})")
```
