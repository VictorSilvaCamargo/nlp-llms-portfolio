```python
from transformers import pipeline
import logging
logging.getLogger("transformers").setLevel(logging.ERROR)

mascarar = pipeline("fill-mask", model="bert-base-multilingual-cased")

texto = "A jogadora marcou o [MASK] [MASK] no segundo tempo"

while "[MASK]" in texto:

    pred = mascarar(texto)
    if isinstance(pred[0], list):
        pred = pred[0]

    texto = texto.replace("[MASK]", pred[0]["token_str"], 1)

resultado = {
    "score": round(pred[0]["score"], 6),
    "token": pred[0]["token"],
    "token_str": pred[0]["token_str"],
    "sequence": texto
}

print(resultado)
```
