```python
import transformers
from transformers import pipeline

#Apenas para suprimir erros, não nescessário.
import logging
logging.getLogger("transformers").setLevel(logging.ERROR)
```

```python
mascarar = pipeline("fill-mask", model="neuralmind/bert-base-portuguese-cased")
texto = mascarar("O goleiro defendeu o [MASK] com uma mão.")
```

```python
for x in range(len(texto)):
  print(texto[x])
```

```python
texto = mascarar("A jogadora marcou o [MASK] no segundo tempo")
```

```python
for x in range(len(texto)):
  print(texto[x])
```
